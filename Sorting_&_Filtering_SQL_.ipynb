Question 8 : Get all employees hired between 2019 and 2022, ordered by HireDate (oldest first).

Ans:

***Query***

- SELECT *
FROM employees_data
WHERE HireDate BETWEEN '2019-01-01' AND '2022-01-01'
ORDER BY HireDate ASC;
