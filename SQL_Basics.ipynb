Q9. What is the Difference Between a Primary Key and a Foreign Key?

Ans:

A **Primary Key** is a unique identifier for each record in a table. It must contain unique values and cannot contain NULL values. Its main purpose is to ensure data integrity and to quickly identify each record.

A **Foreign Key**, on the other hand, is a field (or collection of fields) in one table that uniquely identifies a row of another table. In simple words, a foreign key is a column or a group of columns in a relational database table that provides a link between data in two tables. It acts as a cross-reference between tables because it refers to the primary key of another table, thus establishing a link between them.

Here's a summary of their key differences:

*   **Purpose:** A primary key identifies unique records within a table. A foreign key links records between two tables.
*   **Uniqueness:** Primary key values must be unique. Foreign key values can be duplicated (as multiple records in the referencing table can link to the same record in the referenced table).
*   **NULL Values:** A primary key cannot contain NULL values. A foreign key can contain NULL values (if the relationship allows it).
*   **Table per usage:** A table can have only one primary key. A table can have multiple foreign keys.

Q10. What Are Constraints in SQL and Why Are They Used?

Ans:

**Constraints in SQL** are rules enforced on data columns of a table. They are used to limit the type of data that can go into a table, ensuring the accuracy and reliability of the data. Constraints can be column level (applied to a single column) or table level (applied to the whole table).

**Why Are They Used?**

Constraints are crucial for maintaining **data integrity** within a database. They help ensure that:

1.  **Data Accuracy:** Only valid data is entered into the database, preventing errors and inconsistencies.
2.  **Data Reliability:** The data stored in the database is trustworthy and can be relied upon for decision-making.
3.  **Data Relationships:** Relationships between tables are properly maintained, especially with foreign keys.
4.  **Business Rules Enforcement:** Specific business rules (e.g., age must be positive, prices cannot be zero) are automatically enforced.

**Commonly Used Constraints:**

*   **NOT NULL:** Ensures that a column cannot have a NULL value.
*   **UNIQUE:** Ensures that all values in a column are different.
*   **PRIMARY KEY:** A combination of NOT NULL and UNIQUE. Uniquely identifies each row/record in a database table.
*   **FOREIGN KEY:** Uniquely identifies a row/record in another database table, establishing a link between two tables.
*   **CHECK:** Ensures that all values in a column satisfy a specific condition.
*   **DEFAULT:** Sets a default value for a column when no value is specified.
*   **INDEX:** Used to create and retrieve data from the database very quickly.
