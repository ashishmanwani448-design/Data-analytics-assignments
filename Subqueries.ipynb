Q2) Find the employees who belong to the department with the highest average salary.

Ans:

***Query***:

    select E.name, D.department_name, D.department_id from
    employee_dataset as E join Department_dataset as D
    on E.department_id = D.department_id
    where E.department_id in
    (select Department_id from employee_dataset
    group by Department_id
    having avg(salary) =
    (select max(avg_salary) from
    (select avg(salary) as avg_salary  from employee_dataset
    group by department_id) as T1));

Q2) List departments where the average salary is above ₹55,000.

Ans:

***Query***


    select department_name, avg(T.salary) as avg_salary
    from (select emp_id, department_id, salary from employee_dataset) as T join department_dataset as D
    on T.department_id = D.department_id
    group by department_name
    having avg_salary > 55000
