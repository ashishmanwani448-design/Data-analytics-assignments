Question 1: What is a null hypothesis (H₀) and why is it important in hypothesis testing?

Ans :    

Null hypothesis (H₀) is a statement that there is no effect, no difference, or no relationship between two or more phenomena or populations. It represents the status quo or the default assumption.


it is important in hypothesis testing because:

1) Foundation for Testing: It provides a baseline against which the alternative hypothesis is compared.

2) Statistical Proof: In hypothesis testing, we don't directly 'prove' the alternative hypothesis. Instead, we gather evidence to reject or fail to reject the null hypothesis. If we reject the null hypothesis, it implies that there is sufficient evidence to support the alternative hypothesis.

3) Error Control: It helps in defining and controlling the types of errors in statistical inference (Type I and Type II errors). The significance level (α) is typically set as the probability of incorrectly rejecting a true null hypothesis (Type I error).

4) Objectivity: It provides a structured and objective approach to making decisions based on data, preventing researchers from solely looking for evidence that confirms their initial beliefs.



Question 2: What does the significance level (α) represent in hypothesis testing?

Ans :

The **significance level (α)**, also known as the alpha level, is a pre-determined threshold in hypothesis testing that represents the probability of making a Type I error. A Type I error occurs when you incorrectly reject a true null hypothesis (H₀).

In simpler terms:

1.  **Probability of Type I Error**: If the null hypothesis is actually true, α is the probability that our test results will lead us to mistakenly reject it.
2.  **Threshold for Decision**: It defines how extreme the observed data must be to consider it statistically significant and warrant the rejection of the null hypothesis. Common significance levels are 0.05 (5%), 0.01 (1%), or 0.10 (10%).
3.  **Critical Region**: The significance level determines the size of the "critical region" (or rejection region) in the sampling distribution. If the calculated p-value (the probability of observing data as extreme as, or more extreme than, the current data, assuming the null hypothesis is true) is less than α, we reject the null hypothesis.

For example, if α = 0.05, it means there is a 5% chance of rejecting a true null hypothesis. Researchers choose an alpha level based on the potential consequences of a Type I error in their specific study.

Question 3: Differentiate between Type I and Type II errors.

Ans:

In hypothesis testing, there are two primary types of errors that can occur when making decisions about the null hypothesis:

### Type I Error (False Positive)

*   **Definition**: A Type I error occurs when you **incorrectly reject a true null hypothesis (H₀)**. In other words, you conclude there is a statistically significant effect or relationship when, in reality, there isn't one.
*   **Symbol**: It is denoted by **α (alpha)**, which is also the significance level of the test. The significance level represents the probability of committing a Type I error.
*   **Consequences**: The consequences depend on the context. For example, in a medical test, a Type I error would be diagnosing a healthy person with a disease (false positive). In legal terms, it's like convicting an innocent person.
*   **Control**: The probability of a Type I error is set by the researcher before conducting the experiment (e.g., α = 0.05 or 0.01). Reducing α decreases the chance of a Type I error, but increases the chance of a Type II error.

### Type II Error (False Negative)

*   **Definition**: A Type II error occurs when you **fail to reject a false null hypothesis (H₀)**. In simpler terms, there *is* a statistically significant effect or relationship, but your test fails to detect it.
*   **Symbol**: It is denoted by **β (beta)**. The power of a test (1 - β) is the probability of correctly rejecting a false null hypothesis.
*   **Consequences**: Again, these are context-dependent. In a medical test, a Type II error would be failing to diagnose a sick person with a disease (false negative). In legal terms, it's like letting a guilty person go free.
*   **Control**: The probability of a Type II error is influenced by several factors, including the sample size, the effect size, and the significance level (α). Increasing the sample size or the effect size (the true difference or relationship) generally decreases β. Reducing α increases β.

### Key Differences Summarized:

| Feature             | Type I Error (α)                          | Type II Error (β)                         |
| :------------------ | :---------------------------------------- | :---------------------------------------- |
| **What happens**    | Reject a true H₀                          | Fail to reject a false H₀                 |
| **Outcome**         | False Positive                            | False Negative                            |
| **Conclusion**      | Say an effect exists when it doesn't      | Say no effect exists when it does         |
| **Probability**     | Significance Level (α)                    | Beta (β)                                  |
| **Related concept** | Level of Significance                     | Power of the test (1 - β)                 |
| **Analogy**         | Convicting an innocent person             | Letting a guilty person go free           |

Question 4: Explain the difference between a one-tailed and two-tailed test. Give an example of each.

Ans:

In hypothesis testing, the terms one-tailed and two-tailed refer to the **directionality of the alternative hypothesis (H₁)** and, consequently, the region where we are looking for evidence to reject the null hypothesis.

### Two-Tailed Test

*   **Definition**: A two-tailed test (or non-directional test) is used when the alternative hypothesis states that there *is* a difference or an effect, but does not specify the direction of that difference (i.e., it could be greater than or less than). The rejection region is split between both tails of the sampling distribution.
*   **Hypotheses Structure**:
    *   **H₀**: No difference (e.g., μ = μ₀)
    *   **H₁**: There is a difference (e.g., μ ≠ μ₀)
*   **Rejection Region**: The critical region is divided into two equal parts at both ends (tails) of the sampling distribution. You reject H₀ if the test statistic falls into either the upper or lower critical region.
*   **When to Use**: When you are interested in detecting *any* deviation from the null hypothesis, regardless of its direction.
*   **Example**: A pharmaceutical company wants to know if a new drug has *any* effect on blood pressure (either increasing or decreasing it) compared to a placebo. They don't have a prior expectation of the direction.
    *   **H₀**: The new drug has no effect on blood pressure (μ_drug = μ_placebo).
    *   **H₁**: The new drug has an effect on blood pressure (μ_drug ≠ μ_placebo).

### One-Tailed Test

*   **Definition**: A one-tailed test (or directional test) is used when the alternative hypothesis specifies a particular direction for the difference or effect (i.e., greater than or less than). The entire rejection region is located in only one tail of the sampling distribution.
*   **Hypotheses Structure**:
    *   **Upper-Tailed Test**: H₀: μ ≤ μ₀; H₁: μ > μ₀
    *   **Lower-Tailed Test**: H₀: μ ≥ μ₀; H₁: μ < μ₀
*   **Rejection Region**: The critical region is entirely in one tail of the sampling distribution (either the upper tail for 'greater than' or the lower tail for 'less than'). You reject H₀ if the test statistic falls into this single critical region.
*   **When to Use**: When you have a specific prior expectation about the direction of the difference or effect based on theory, previous research, or practical considerations.
*   **Example (Upper-Tailed)**: A company implements a new training program for its sales team and wants to know if it *increases* the average sales performance.
    *   **H₀**: The new training program does not increase sales performance (μ_after ≤ μ_before).
    *   **H₁**: The new training program increases sales performance (μ_after > μ_before).
*   **Example (Lower-Tailed)**: A quality control manager wants to test if a new manufacturing process *reduces* the average number of defects per batch.
    *   **H₀**: The new process does not reduce defects (μ_new ≥ μ_old).
    *   **H₁**: The new process reduces defects (μ_new < μ_old).

### Key Differences Summarized:

| Feature           | Two-Tailed Test                               | One-Tailed Test                               |
| :---------------- | :-------------------------------------------- | :-------------------------------------------- |
| **H₁ Direction**  | Non-directional (e.g., ≠)                     | Directional (e.g., >, <)                      |
| **Rejection Region** | Split into two tails                          | Located entirely in one tail                  |
| **Interest**      | Any difference from H₀                        | Difference in a specific direction            |
| **α Placement**   | α/2 in each tail                              | α entirely in one tail                        |
| **Power**         | Generally lower power to detect a specific directional effect (for a given α) | Generally higher power to detect a specific directional effect (for a given α) |

Choosing between a one-tailed and two-tailed test depends critically on the research question and the prior knowledge or hypothesis about the direction of the effect.

Question 6: When should you use a Z-test instead of a t-test?

Ans:

Both Z-tests and t-tests are statistical hypothesis tests used to compare means. The primary difference between them lies in the **information available about the population standard deviation** and the **sample size**.

### Use a Z-test when:

1.  **Population Standard Deviation (σ) is Known**: The most crucial condition for using a Z-test is that you know the true standard deviation of the population from which your sample is drawn. This is often the case in quality control or when extensive historical data is available.
2.  **Large Sample Size (n ≥ 30)**: Even if the population standard deviation is unknown, if the sample size is large (generally n ≥ 30), the sample standard deviation (s) is often a good estimate of the population standard deviation (σ). In such cases, the sampling distribution of the sample mean approaches a normal distribution due to the Central Limit Theorem, making the Z-test applicable.
3.  **Population is Normally Distributed**: If the population is known to be normally distributed, a Z-test can be used for any sample size, provided the population standard deviation is known.

### Use a t-test when:

1.  **Population Standard Deviation (σ) is Unknown**: This is the most common scenario in real-world research. When you don't know the population standard deviation, you must estimate it from your sample using the sample standard deviation (s). This introduces more variability, and the t-distribution (which has fatter tails than the normal distribution) accounts for this increased uncertainty.
2.  **Small Sample Size (n < 30)**: When the sample size is small, the sample standard deviation may not be a reliable estimate of the population standard deviation. The t-distribution is specifically designed for situations with small sample sizes and unknown population standard deviations, providing more conservative (wider) confidence intervals and critical values to account for the greater uncertainty.
3.  **Population is Approximately Normally Distributed**: A t-test assumes that the population from which the sample is drawn is approximately normally distributed. If the sample size is very small and the population is highly non-normal, neither a Z-test nor a t-test might be appropriate, and non-parametric tests might be considered.

### Summary Table:

| Feature                         | Z-test                                    | t-test                                       |
| :------------------------------ | :---------------------------------------- | :------------------------------------------- |
| **Population Std. Dev. (σ)**    | Known                                     | Unknown (estimated by sample std. dev. `s`)  |
| **Sample Size (n)**             | Large (n ≥ 30) or any size if σ is known and population is normal | Small (n < 30)                               |
| **Distribution Used**           | Z-distribution (Standard Normal)          | t-distribution (Student's t-distribution)    |
| **Impact on Critical Values**   | Critical values are fixed                 | Critical values vary with degrees of freedom |
| **Conservativeness**            | Less conservative (narrower intervals)    | More conservative (wider intervals)          |

Question 7: The productivity of 6 employees was measured before and after a training program. At α = 0.05, test if the training improved productivity.

Ans:



### Given:

Productivity of 6 employees was measured **before** and **after** training.

| Employee | Before | After | Difference (After − Before) |
| -------- | ------ | ----- | --------------------------- |
| 1        | 50     | 55    | 5                           |
| 2        | 60     | 65    | 5                           |
| 3        | 58     | 59    | 1                           |
| 4        | 55     | 58    | 3                           |
| 5        | 62     | 63    | 1                           |
| 6        | 56     | 59    | 3                           |

Significance level:
[alpha = 0.05]

---

### Step 1: Hypotheses

Since we want to test whether training **improved** productivity:

* **Null Hypothesis (H₀):**
  [d = 0]

* **Alternative Hypothesis (H₁):**
  [d > 0]

(This is a **one-tailed paired t-test**)

---

### Step 2: Mean of differences

[
\bar{d} = \frac {5+5+1+3+1+3}{6} = \frac{18}{6} = 3
]

---

### Step 3: Standard deviation of differences

[
\sum (d - \bar{d})^2 = 16
]

[
s_d = \sqrt{\frac{16}{6-1}} = \sqrt{3.2} = 1.789
]

---

### Step 4: Test statistic

[
t = \frac{\bar{d}}{s_d / \sqrt{n}}
]

[
t = \frac{3}{1.789/\sqrt{6}}
]

[
t = 4.11
]

---

### Step 5: Critical value

Degrees of freedom:
[
df = n - 1 = 5
]

From t-table at ( \alpha = 0.05 ) (one-tailed):

[
t_{critical} = 2.015
]

---

### Step 6: Decision

[
t_{calculated} > t_{critical}
]

[
4.11 > 2.015
]

✅ **Reject H₀**

---

### Step 7: Conclusion

At the 5% level of significance, there is **sufficient evidence** to conclude that the **training program significantly improved employee productivity**.
