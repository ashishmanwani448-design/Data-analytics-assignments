## Aggregation

Question 1 : Count how many cities are there in each country?

Ans:

***Query***:

    "select distinct co.name as country,
       count(c.id) as cities
    from city as c join Country as Co
    on c.countrycode = Co.code
    group by Co.name order by count(c.id) desc;

Question 2 : Display all continents having more than 30 countries.

Ans:

***Query***:

    select distinct continent, count(name)
    from Country
    group by continent
    having count(name) > 30
    order by count(name) desc;


***Result***:

/content/question2.csv

Question 3 : List regions whose total population exceeds 200 million.

Ans:



***Query***:

    select distinct Region, sum(population) as total_population
    from Country
    group by Region
    having sum(population) > 200000000
    order by sum(population) desc;


***Result***:

/content/question3.csv

Question 4 : Find the top 5 continents by average GNP per country.

Ans:


***Query***:

    select distinct name as country, avg(GNP) as average_GNP
    from Country
    group by name
    order by  avg(GNP) desc
    limit 5;

***Result***:

/content/question4.csv

Question 5 : Find the total number of official languages spoken in each continent.


Ans:

***Query***:
    
    SELECT
    C.continent AS continent,
    COUNT(DISTINCT Cl.language) AS total_languages
    FROM Country AS C
    JOIN CountryLanguage AS Cl
    ON C.code = Cl.countrycode
    WHERE Cl.IsOfficial = 'T'
    GROUP BY C.continent;



***Result***:

/content/question5.csv

Question 7 : Find the country with the highest average city population.

Ans:


***Query***:

    from country as CO join City AS CI
